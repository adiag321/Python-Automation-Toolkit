```python
import os
import pandas as pd
import json
```

```python
# Set the path to your Chrome bookmarks file
bookmarks_path = os.path.join(os.environ['HOME'], 'Library', 'Application Support', 'Google', 'Chrome', 'Default', 'Bookmarks')
bookmarks_path
```

```python
# Load the bookmarks file
with open(bookmarks_path, 'r') as f:
    bookmarks = json.load(f)
```

```python
# Set the name of the folder you're interested in
folder_name = 'SQL'


# Initialize an empty list to store the bookmarks
bookmarks_list = []
# Function to recursively traverse the bookmarks tree
def traverse_bookmarks(bookmarks, folder_name):
    for bookmark in bookmarks:
        if bookmark['type'] == 'folder' and bookmark['name'] == folder_name:
            for child in bookmark['children']:
                if 'url' in child:
                    bookmarks_list.append({
                        'Title': child['name'],
                        'URL': child['url']
                    })
        elif 'children' in bookmark:
            traverse_bookmarks(bookmark['children'], folder_name)
```

```python
# Traverse the bookmarks tree
traverse_bookmarks(bookmarks['roots']['bookmark_bar']['children'], folder_name)

# Create a pandas dataframe from the bookmarks list
df = pd.DataFrame(bookmarks_list)
df
```
